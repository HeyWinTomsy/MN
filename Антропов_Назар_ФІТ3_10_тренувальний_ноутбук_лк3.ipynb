## Завдання 3. Скалярний добуток і норма вектора

Для векторів `u = [3, -1, 2]` та `v = [1, 4, -2]`:

1. Обчисліть скалярний добуток.
2. Знайдіть евклідову норму кожного вектора.
3. Обчисліть косинус кута між векторами за формулою

\[
\cos(\alpha)=\frac{u\cdot v}{\|u\|\|v\|}
\]


````python
u = np.array([3, -1, 2])
v = np.array([1, 4, -2])

# Ваш код тут
dobutok = np.dot(u, v)
nu = np.linalg.norm(u)
nv = np.linalg.norm(v)
cos_a = dobutok / (nu * nv)

print("Скалярний добуток:", dobutok)
print("Норма u:", nu)
print("Норма v:", nv)
print("Косинус кута:", cos_a)
````

## Завдання 8. Визначник та обернена матриця

Для матриці

\[
A =
\begin{bmatrix}
4 & 7\\
2 & 6
\end{bmatrix}
\]

1. Знайдіть визначник.
2. Знайдіть обернену матрицю.
3. Перевірте результат множенням `A @ A_inv`.
4. Поясніть у коментарі, який результат має бути отриманий.

````python
A = np.array([[4, 7],
              [2, 6]])

# Ваш код тут
det_A = np.linalg.det(A)
print("Визначник A:", det_A)

A_inv = np.linalg.inv(A)
print("\nОбернена матриця A:")
print(A_inv)

result = A @ A_inv
print("\nA і A_inv:")
print(result)
````

## Завдання 13. Функція з умовою

Створіть функцію

\[
f(x) =
\begin{cases}
x^2, & x < 0 \\
2x + 1, & x \geq 0
\end{cases}
\]

Застосуйте її до вектора значень від `-5` до `5`.

Спробуйте реалізувати функцію двома способами:
1. через звичайну функцію Python;
2. через `np.where()`.

````python
x = np.arange(-5, 6)
# Варіант 1
def f(x):
    if x < 0:
        return x ** 2
    else:
        return 2 * x + 1

result = np.array([f(i) for i in x])

print("x =", x)
print("f(x) =", result)

# Варіант 2
result = np.where(x < 0, x ** 2, 2 * x + 1)

print("x =", x)
print("f(x) =", result)
````

## Завдання 15. Підсумкове

Згенеруйте матрицю `5 × 5` із випадкових цілих чисел від `1` до `100`.

Виконайте такі дії:
1. Знайдіть середнє значення кожного рядка.
2. Знайдіть середнє значення кожного стовпця.
3. Замініть усі елементи, менші за загальне середнє матриці, на `0`.
4. Знайдіть рядок із найбільшою сумою.
5. Нормалізуйте початкову матрицю за формулою:

\[
x_{norm} = \frac{x - x_{min}}{x_{max} - x_{min}}
\]

6. Напишіть функцію, яка виконує таку нормалізацію для довільної числової матриці.

````python
rng = np.random.default_rng(42)
matrix = rng.integers(1, 101, size=(5, 5))

print("Початкова матриця:")
print(matrix)

# Ваш код тут

row_mean = np.mean(matrix, axis=1)
print("\nСереднє значення кожного рядка:")
print(row_mean)

column_mean = np.mean(matrix, axis=0)
print("\nСереднє значення кожного стовпця:")
print(column_mean)

mean = np.mean(matrix)

matrix_zero = matrix.copy()
matrix_zero[matrix_zero < mean] = 0

print("\nЗагальне середнє:", mean)
print("Матриця після заміни:")
print(matrix_zero)

row_sums = np.sum(matrix, axis=1)
max_row = np.argmax(row_sums)

print("\nСуми рядків:")
print(row_sums)
print("Рядок із найбільшою сумою:", max_row + 1)

x_min = np.min(matrix)
x_max = np.max(matrix)

normalized = (matrix - x_min) / (x_max - x_min)

print("\nНормалізована матриця:")
print(normalized)

def normalize_matrix(matrix):
    x_min = np.min(matrix)
    x_max = np.max(matrix)

    if x_max == x_min:
        return np.zeros_like(matrix, dtype=float)

    return (matrix - x_min) / (x_max - x_min)

result = normalize_matrix(matrix)

print("\nРезультат роботи функції:")
print(result)
````
